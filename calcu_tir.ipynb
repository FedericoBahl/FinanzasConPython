```python
# Filtro el DF por tipo de ticker

df_al29 = df[df['Ticker'] == 'AL29D']
df_al30 = df[df['Ticker'] == 'AL30D']
df_al35 = df[df['Ticker'] == 'AL35D']
df_ae38 = df[df['Ticker'] == 'AE38D']
df_al41 = df[df['Ticker'] == 'AL41D']

```

```python
# Definición de la función

def calculo_tir2():
  bono = entry_bono.get().upper()
  precio_hoy = entry_precio.get()


  if bono == 'AL29D' or bono == 'GD29D':
    inversion_inicial = pd.DataFrame({'Ticker': [bono], 'Fecha': [date.today()], 'R+A': [float(precio_hoy)*-1]})
    flujos = pd.concat([inversion_inicial, df_al29])

    tasa_retorno = xirr(flujos['Fecha'], flujos['R+A'])

    label_resultado.config(text=f'La TIR del bono {bono} es: {tasa_retorno*100: .2f}%', fg='blue')

  elif bono == 'AL30D' or bono == 'GD30D':
    inversion_inicial = pd.DataFrame({'Ticker': [bono], 'Fecha': [date.today()], 'R+A': [float(precio_hoy)*-1]})
    flujos = pd.concat([inversion_inicial, df_al30])

    tasa_retorno = xirr(flujos['Fecha'], flujos['R+A'])

    label_resultado.config(text=f'La TIR del bono {bono} es: {tasa_retorno*100: .2f}%', fg='blue')

  elif bono == 'AL35D' or bono == 'GD35D':
    inversion_inicial = pd.DataFrame({'Ticker': [bono], 'Fecha': [date.today()], 'R+A': [float(precio_hoy)*-1]})
    flujos = pd.concat([inversion_inicial, df_al35])

    tasa_retorno = xirr(flujos['Fecha'], flujos['R+A'])

    label_resultado.config(text=f'La TIR del bono {bono} es: {tasa_retorno*100: .2f}%', fg='blue')

  elif bono == 'AE38D' or bono == 'GD38D':
    inversion_inicial = pd.DataFrame({'Ticker': [bono], 'Fecha': [date.today()], 'R+A': [float(precio_hoy)*-1]})
    flujos = pd.concat([inversion_inicial, df_ae38])

    tasa_retorno = xirr(flujos['Fecha'], flujos['R+A'])

    label_resultado.config(text=f'La TIR del bono {bono} es: {tasa_retorno*100: .2f}%', fg='blue')

  elif bono == 'AL41D' or bono == 'GD41D':
    inversion_inicial = pd.DataFrame({'Ticker': [bono], 'Fecha': [date.today()], 'R+A': [float(precio_hoy)*-1]})
    flujos = pd.concat([inversion_inicial, df_al41])

    tasa_retorno = xirr(flujos['Fecha'], flujos['R+A'])

    label_resultado.config(text=f'La TIR del bono {bono} es: {tasa_retorno*100: .2f}%', fg='blue')

  else:
    label_resultado.config(text=f'Bono no encontrado', fg='red')

```

```python
# Ventana de ejecución del programa

root = tk.Tk()
root.title('Calculadora de TIR de Bonos Soberanos')
root.geometry('500x250')
fecha_hoy = date.today().strftime("%d/%m/%Y")

tk.Label(root, text=fecha_hoy).pack(pady=5)


### Etiqueta de la instrucción
tk.Label(root, text='Ticker del bono (ej: AL30D, GD38D):').pack(pady=5)
entry_bono = tk.Entry(root)
entry_bono.pack()

tk.Label(root, text='Precio del bono en USD:').pack(pady=5)
entry_precio = tk.Entry(root)
entry_precio.pack()

### Botón para ejecutar el calculo
boton_calcular = tk.Button(root, text='Calcular TIR', command=calculo_tir2)
boton_calcular.pack(pady=20)

### Mostrar el resultado
label_resultado = tk.Label(root, text='', font=("Tahoma", 12, 'bold'))
label_resultado.pack()

### Iniciar bucle de la aplicación

root.mainloop()
```
